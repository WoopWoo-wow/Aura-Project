# <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Introduction</p>

<p style="text-align: justify;">The <strong>Student Sleep Patterns Dataset</strong> explores the sleep habits of university students, aiming to identify how various lifestyle factors influence sleep duration and quality. Designed with synthetic yet realistic data, this dataset mimics plausible distributions and relationships, making it an excellent resource for educational and research purposes. The dataset captures various attributes such as sleep schedules, study hours, screen time, caffeine consumption, and physical activity to provide a comprehensive view of student sleep patterns.</p>

<p style="text-align: justify;">This dataset includes 500 entries, each representing a unique student, with 14 variables offering insights into demographic details, lifestyle habits, and subjective sleep quality ratings. With its realistic design, the dataset serves as a valuable tool for investigating correlations, building predictive models, and visualizing patterns to enhance understanding of sleep behaviors in student populations.</p>

```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import kruskal,probplot
import statsmodels.api as sm
from sklearn.inspection import permutation_importance
from sklearn.ensemble import RandomForestRegressor
import shap

import warnings
warnings.filterwarnings('ignore')

cmap = matplotlib.colormaps.get_cmap('rocket_r')
```

# <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Data Overview</p>

- **Total Rows:** 500 (students)  
- **Total Columns:** 14 (attributes related to demographics, sleep, and lifestyle)  
- **Key Variables:**  
    - **Student_ID**: Unique identifier for each student (integer).
    - **Age**: Age of the student in years (integer).
    - **Gender**: Gender of the student (categorical: 'Male', 'Female', 'Other').
    - **University_Year**: The student's current year of university (categorical: '1st Year', '2nd Year', '3rd Year', '4th Year').
    - **Sleep_Duration**: Total hours of sleep per night (float).
    - **Study_Hours**: Average number of hours spent studying per day (float).
    - **Screen_Time**: Average number of hours spent on screens (excluding studying) per day (float).
    - **Caffeine_Intake**: Average number of caffeinated beverages consumed per day (integer).
    - **Physical_Activity**: Average minutes spent on physical activity per day (integer).
    - **Sleep_Quality**: Subjective rating of sleep quality on a scale of 1 to 10 (1 being the worst, 10 being the best) (integer).
    - **Weekday_Sleep_Start**: Time the student typically goes to sleep on weekdays (float, in 24-hour format, e.g., 23.00 is 11 PM).
    - **Weekend_Sleep_Start**: Time the student typically goes to sleep on weekends (float, in 24-hour format).
    - **Weekday_Sleep_End**: Time the student typically wakes up on weekdays (float, in 24-hour format).
    - **Weekend_Sleep_End**: Time the student typically wakes up on weekends (float, in 24-hour format).

# <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Exploratory Data Analysis</p>

```python
# Sleep Duration on Weekdays and Weekends
data['Weekday_Sleep_Duration'] = data['Weekday_Sleep_End'] - data['Weekday_Sleep_Start']
data['Weekend_Sleep_Duration'] = data['Weekend_Sleep_End'] - data['Weekend_Sleep_Start']

# Adjust for any negative values (e.g., if sleep starts late at night and ends after midnight)
data['Weekday_Sleep_Duration'] = data['Weekday_Sleep_Duration'].apply(lambda x: x + 24 if x < 0 else x)
data['Weekend_Sleep_Duration'] = data['Weekend_Sleep_Duration'].apply(lambda x: x + 24 if x < 0 else x)
```

```python
#Average Sleep Time
data['Avg_Sleep_Duration'] = (data['Weekday_Sleep_Duration'] + data['Weekend_Sleep_Duration']) / 2
```

```python
# Difference in Sleep Start Time
data['Sleep_Start_Difference'] = data['Weekend_Sleep_Start'] - data['Weekday_Sleep_Start']

# Difference in Sleep End Time
data['Sleep_End_Difference'] = data['Weekend_Sleep_End'] - data['Weekday_Sleep_End']
```

```python
# Categorize time into periods
def sleep_period(time):
    if 4 <= time < 6:
        return 'Early Morning'
    elif 6 <= time < 12:
        return 'Morning'
    elif 12 <= time < 18:
        return 'Afternoon'
    elif 18 <= time < 24:
        return 'Night'
    return 'Late Night'

data['Weekday_Sleep_Start_Period'] = data['Weekday_Sleep_Start'].apply(sleep_period)
data['Weekend_Sleep_Start_Period'] = data['Weekend_Sleep_Start'].apply(sleep_period)

data['Weekday_Sleep_End_Period'] = data['Weekday_Sleep_End'].apply(sleep_period)
data['Weekend_Sleep_End_Period'] = data['Weekend_Sleep_End'].apply(sleep_period)
```

```python
discrete_feat = ['Age', 'Gender', 'University_Year', 'Caffeine_Intake','Weekday_Sleep_Start_Period', 'Weekend_Sleep_Start_Period', 'Weekday_Sleep_End_Period', 'Weekend_Sleep_End_Period']
continuous_feat = [col for col in data.select_dtypes(exclude='O') if col not in ['Sleep_Quality']+discrete_feat]

print('Discrete Features:',discrete_feat)
print('Continuous Features:',continuous_feat)
```

## <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Discrete Features Analysis</p>

```python
for feat in ['Sleep_Quality']+discrete_feat:
    agg_data = data.groupby(feat).size()
    norm = plt.Normalize(agg_data.min(), agg_data.max())
    
    plt.figure(figsize=(8,3))
    
    plt.subplot(1,2,1)
    plt.pie(agg_data,
            autopct='%.2f%%',
            labels=agg_data.index,
            colors=sns.color_palette("rocket_r", len(agg_data)+2),
            radius=1.25)
    plt.pie([1],radius=0.65,colors=['white'])
    
    plt.subplot(1,2,2)
    bars = sns.barplot(x=agg_data.index,y=agg_data)
    bars.bar_label(bars.containers[0], 
                   rotation=90,fontsize=10,
                   label_type='center',
                   labels=[f'{x}' for x in agg_data])
    
    for bar in bars.patches:
        height = bar.get_height()
        bar.set_facecolor(cmap(norm(height)*0.75))
    
    plt.legend([])
    plt.yticks([])
    plt.ylabel('')
    plt.xlabel('')
    plt.box(False)
    
    plt.suptitle(feat+' Distribution')
    plt.tight_layout()
    plt.show()
```

<p class="alert-info" style="padding:10px 20px">The features are fairly evenly distributed.</p>

```python
for feat in discrete_feat:
    groups = data.groupby(feat)['Sleep_Quality'].apply(list)
    if len(groups)>1:
        _, p_value = kruskal(*groups)
    else:
        p_value = 0.0
    
    plt.figure(figsize=(8,3))
    plt.subplot(121)
    crosstab = data.groupby(['Sleep_Quality',feat]).size().unstack()
    crosstab.plot(kind='bar',stacked=True,cmap='rocket',ax=plt.gca())
    plt.box(False)

    plt.subplot(122)
    sns.lineplot(data=data.groupby(feat)['Sleep_Quality'].mean(),color='tomato',linewidth=3,marker='o',markersize=10)
    plt.box(False)
    
    plt.suptitle(f'Sleep Quality by {feat} [P Value :{p_value:0.2f}]')
    plt.tight_layout()
    plt.show()
```

<p class="alert-info" style="padding:10px 20px">The discrete features do not show a statistically significant relationship with sleep quality. </p>

## <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Continuous Features Analysis</p>

```python
for feat in continuous_feat:
    counts, bins = np.histogram(data[feat].dropna(), bins=50)
    norm = plt.Normalize(counts.mean()-counts.std(), counts.max())
    
    plt.figure(figsize=(8,3))
    
    plt.subplot(1,2,1)
    hist = sns.histplot(x=data[feat].dropna(),bins=50,kde=True)
    for patch in hist.patches:
        height = patch.get_height()  
        patch.set_facecolor(cmap(norm(height)*0.75))
    plt.box(False)
    plt.ylabel('')
    
    plt.subplot(1,2,2)
    probplot(data[feat],plot=plt)
    plt.box(False)
    plt.ylabel('')
    
    plt.suptitle(feat+' Distribution')
    plt.tight_layout()
    plt.show()
```

```python
for feat in continuous_feat:
    corr = data[['Sleep_Quality']].corrwith(data[feat],method='spearman')
    plt.figure(figsize=(8,3))
    
    plt.subplot(121)
    sns.kdeplot(data=data, x='Sleep_Quality', y=feat, fill=True,levels=20, cmap="rocket_r")
    plt.box(False)

    plt.subplot(122)
    sns.boxplot(data=data,x='Sleep_Quality',y=feat,palette='rocket_r')
    plt.box(False)
    
    plt.suptitle(f'Sleep_Quality X {feat} [Correlation:{corr.values[0]:0.2f}]')
    plt.tight_layout()
    plt.show()
    
```

<p class="alert-info" style="padding:10px 20px">The continuous features do not show a significant correlation with sleep quality.</p>

# <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Multiple Regression Analysis</p>

In our previous attempt to identify statistically significant features influencing Sleep Quality, we faced some challenges. Now, let's focus on finding the most important predictors for Sleep Quality using a more reliable approach—Multiple Regression Analysis.

```python
data['Gender'] = data['Gender'].replace({'Male':1,'Female':2,'Other':3}).astype(int)
data['University_Year'] = data['University_Year'].replace({'1st Year':1,'2nd Year':2,'3rd Year':3,'4th Year':4}).astype(int)
data['Weekday_Sleep_Start_Period'] = data['Weekday_Sleep_Start_Period'].replace({ 'Early Morning': 0, 'Morning': 1, 'Afternoon': 2, 'Night': 3, 'Late Night': 4 }).astype(int)
data['Weekend_Sleep_Start_Period'] = data['Weekend_Sleep_Start_Period'].replace({ 'Early Morning': 0, 'Morning': 1, 'Afternoon': 2, 'Night': 3, 'Late Night': 4 }).astype(int)

data['Weekday_Sleep_End_Period'] = data['Weekday_Sleep_End_Period'].replace({ 'Early Morning': 0, 'Morning': 1, 'Afternoon': 2, 'Night': 3, 'Late Night': 4 }).astype(int)
data['Weekend_Sleep_End_Period'] = data['Weekend_Sleep_End_Period'].replace({ 'Early Morning': 0, 'Morning': 1, 'Afternoon': 2, 'Night': 3, 'Late Night': 4 }).astype(int)
```

```python
plt.figure(figsize=(8,8))
sns.heatmap(data.corr(method='spearman'),annot=True, cmap='rocket',cbar=False,fmt='0.2f')
plt.show()
```

```python
X = data.drop(['Sleep_Quality','Sleep_Start_Difference','Sleep_End_Difference','Weekend_Sleep_End_Period','Avg_Sleep_Duration'],axis=1)
y= data['Sleep_Quality']

model = sm.OLS(y,X).fit()
model.summary()
```

<p class="alert-info" style="padding:10px 20px">As per the Regression Anlaysis <strong>Age, Gender</strong> and  <Strong>Study_Hour</strong> are greatest factor in <strong>Sleep_Quality</strong>.</p>

# <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Random Forest Permutation Importance</p>

Now, we will shift our approach and leverage the power of Random Forest along with permutation importance to identify the most influential predictors for Sleep Quality.

```python
X, y = data.drop('Sleep_Quality',axis=1),data['Sleep_Quality']


model = RandomForestRegressor(n_estimators=1000,random_state=123)
model.fit(X,y)

results = permutation_importance(model, X, y,random_state=123)

imp_df = pd.DataFrame({
    'Importance': results.importances_mean
},index=X.columns).sort_values(by='Importance',ascending=False)
sns.barplot(data=imp_df,x='Importance',y=imp_df.index,palette='rocket_r')
```

<p class="alert-info" style="padding:10px 20px">The Random Forest feature importance scores, derived through permutation, yield results that align closely with findings from existing studies.</p>

## <p style="font-size: 36px;background-color:#DC143C; color:white; padding:10px 20px; width:95%; border-radius:15px; text-align:center;margin: 0 auto;font-weight: bold;">Random Forest Importance with SHAP</p>

Let's explain the feature importance

```python
# Create a SHAP explainer
explainer = shap.TreeExplainer(model)

# Compute SHAP values for the train set
shap_values = explainer.shap_values(X)

# Visualize the SHAP values for a summary plot
shap.summary_plot(shap_values, X)
```
